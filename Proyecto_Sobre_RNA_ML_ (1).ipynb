```python
from __future__ import division

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import KFold
from sklearn.preprocessing import OneHotEncoder
import time

def oneHotEncoder(y):
    # Se cambia 'sparse=False' por 'sparse_output=False' debido a la versión de scikit-learn
    onehot_encoder = OneHotEncoder(sparse_output=False)
    integer_encoded = y.reshape(len(y), 1)
    onehot_encoded = onehot_encoder.fit_transform(integer_encoded)
    return onehot_encoded

# cargamos la bd de entrenamiento desde la ruta local correcta
db = np.loadtxt('DB_Fetal_Cardiotocograms.txt', delimiter='	')

X=db[:,0:22]

Y = db[:,22]
Y = oneHotEncoder(Y)
print(Y)
print(X)
```

## Ejercicio 1.

1.1. Por qué es necesario hacer la codificación One-Hot Encodign de la variable de salida?

R/: La variable de salida (NSP) es categórica nominal: 1 = Normal, 2 = Sospechoso y 3 = Patológico son etiquetas sin orden ni distancia entre ellas. Si se dejaran como números 1, 2 y 3, la red las interpretaría como valores ordenados (por ejemplo, que "3" es el triple de "1"), lo cual no tiene sentido. Con One-Hot cada clase se representa como un vector con un 1 en la posición de su clase y 0 en las demás ([1,0,0], [0,1,0], [0,0,1]). Así la capa de salida tiene una neurona por clase, cuya salida (softmax) se interpreta como la probabilidad de pertenecer a esa clase, y la función de pérdida (entropía cruzada categórica) puede comparar ese vector de probabilidades con el vector real para calcular el error y ajustar los pesos en el entrenamiento supervisado.

1.2 De qué manera se toma la decisión para asignar la clase una muestra en la red feed-forward cuando hay múltiples clases?

R/: La capa de salida tiene una neurona por clase con función de activación softmax, que convierte las salidas en probabilidades que suman 1. La muestra se asigna a la clase cuya neurona tenga la mayor probabilidad (argmax). Es lo que hace la función `metrics`: compara `np.argmax(test)` con `np.argmax(pred)` y cuenta un error cuando no coinciden.


La siguiente celda tiene un método para calcular las medidas de error, segúún sea un problema biclase o multiclase, solo debe ejecutarla.

```python
def metrics(y_test, y_pred):
  '''
  Parametros:
    - lista (array) y_test (One-hot encoding)
    - lista (array) y_predichas (softmax vector)
  Return:
    - F1
    - Recall
    - Precision
    - Accuracy
    - Especificity
  '''
  y_test = np.asarray(y_test)
  y_pred = np.asarray(y_pred)

  if y_test[0].shape[0] == 1:    #Biclase
    TP,TN,FP,FN = 0,0,0,0
    for t,p in zip(y_test,y_pred):
      if t==1:
        if p>=0.5: #yp==1:
          TP+=1
        else:
          FN+=1
      else:
        if p>=0.5: #yp==1:
          FP+=1
        else:
          TN+=1

    recall = TP/(TP+FN)
    prec = TP/(TP+FP)
    acc = (TP+TN)/(TP+FP+FN+TN)
    f1 = (2*recall*prec)/(recall+prec)
    esp = TN/(TN+FP)

    return f1, recall, prec, acc, esp

  else:    #Multiclase
    error = 0
    for test,pred in zip(y_test,y_pred):
        t = np.argmax(test)
        p = np.argmax(pred)
        if t!=p:
          error+=1

    err = error/np.shape(y_test)[0]
    acc = 1 - err

    return acc
```

```python
import keras
from keras.models import Sequential
from keras.layers import Dense, Dropout, Activation
from sklearn.model_selection import KFold
import time
import numpy as np

#Para calcular el costo computacional
tiempo_i = time.time()

# Número de características de entrada (22 columnas en X)
input_dim = X.shape[1]
# Número de clases de salida (3 clases: Normal, Suspect, Pathologic)
output_dim = Y.shape[1]

# Creamos la red neuronal artificial secuencial
model = Sequential()
model.add(Dense(10, input_dim=input_dim, activation='relu')) # Capa oculta con 10 neuronas
model.add(Dense(2, input_dim=input_dim, activation='relu')) # Capa oculta con 2 neuronas
model.add(Dense(output_dim, activation='softmax'))           # Capa de salida con activación Softmax

#Implemetamos la metodología de validación cross validation con 10 folds
Accuracy = np.zeros(10)

kf = KFold(n_splits=10, shuffle=True, random_state=42)
j = 0
for train_index, test_index in kf.split(X):
    X_train, X_test = X[train_index], X[test_index]
    y_train, y_test = Y[train_index], Y[test_index]

    # Compilamos el modelo usando pérdida de entropía cruzada categórica y el optimizador adam
    model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])

    # Entrenamos el modelo durante 10 épocas (ajustable según la tabla)
    model.fit(X_train, y_train, epochs=50, batch_size=32, verbose=0)

    # Validación del modelo haciendo predicciones sobre el conjunto de test
    y_pred = model.predict(X_test, verbose=0)

    Accuracy[j] = metrics(y_test, y_pred)
    j+=1

print("\nEficiencia en la clasificación: " + str(np.mean(Accuracy)) + " +/- " + str(np.std(Accuracy)))

print ("\n\nTiempo total de ejecución: " + str(time.time()-tiempo_i)+ " segundos.")
```

Ahora va a resolver el problema de regresión de la base de datos AirQuality (https://archive.ics.uci.edu/ml/datasets/Air+Quality) usando RNA bajo la arquitectura feed-forward. En la siguiente celda no tiene que implementar código, solo ejecutarla y analizar que proceso de limpieza que se lleva a cabo.

```python
import numpy as np
import os
import urllib.request
import zipfile

# Descargar y descomprimir el dataset si no existe localmente
if not os.path.exists('AirQuality.data'):
    print("Descargando el dataset Air Quality...")
    url = 'https://archive.ics.uci.edu/static/public/360/air+quality.zip'
    urllib.request.urlretrieve(url, 'AirQuality.zip')
    with zipfile.ZipFile('AirQuality.zip', 'r') as zip_ref:
        zip_ref.extract('AirQuality.csv', path='.')

    # Convertir el archivo CSV original (delimitado por ';') al formato esperado con tabulaciones
    # y reemplazamos comas decimales por puntos.
    with open('AirQuality.csv', 'r', encoding='utf-8', errors='ignore') as f_in:
        lines = f_in.readlines()

    with open('AirQuality.data', 'w', encoding='utf-8') as f_out:
        for line in lines[1:9358]: # Tomar solo los 9357 registros de datos válidos
            parts = line.strip().split(';')
            if len(parts) >= 15:
                # Seleccionar las columnas numéricas que se usarán (de la 2 a la 14 en la estructura del dataset)
                selected = parts[2:15]
                # Reemplazar comas por puntos para que numpy pueda interpretarlo como float
                selected = [val.replace(',', '.') if val else '-200' for val in selected]
                f_out.write('\t'.join(selected) + '\n')
    print("Dataset descargado y procesado como 'AirQuality.data'.\n")

# cargamos la bd de entrenamiento desde la ruta local
db = np.loadtxt('AirQuality.data', delimiter='\t')
print("Dim de la base de datos original: " + str(np.shape(db)))
db = db.reshape(9357, 13)
DataBase = db

# La base de datos AirQuality tiene muchos valores perdidos, vamos a quitar todos esos registros de la base de datos
# Están marcados como -200. Donde haya un valor -200 eliminaremos el registro.
j = 0
for i in range(0, np.size(db, 0)):
    if -200 == db[i, 12]:
        j += 1
        DataBase = np.delete(DataBase, i, 0)

print("\nHay " + str(j) + " valores perdidos en la variable de salida. :(")
print("\nDim de la base de datos sin las muestras con variable de salida perdido " + str(np.shape(DataBase)))

# Ya hemos eliminado los registros con valor de la variable de salida perdido.

# Ahora vamos a imputar los valores perdidos en cada una de las características
print("\nProcesando imputación de valores perdidos en las características . . .\n")

for k in range(0, np.size(DataBase, 0)):
    for w in range(0, 13):
        if -200 == DataBase[k, w]:
            DataBase[k, w] = round(np.mean(DataBase[:, w]))

print("Imputación finalizada.\n")

hay_missed_values = False
for i in range(0, np.size(DataBase, 0)):
    if -200 in DataBase[i, :]:
        hay_missed_values = True
if hay_missed_values:
    print("Hay valores perdidos")
else:
    print("No hay valores perdidos en la base de datos. Ahora se puede procesar. La base de datos está en la variable DataBase")
```

```python
X = DataBase[:,0:12]

Y = DataBase[:,12]*10 #(se multiplica por 100 para evitar errores en la interpretación de la eficiencia del modelo porque
                      #los valores de Y menores que 1 pueden llevar a malas interpretaciones).
Y = Y.reshape(np.size(X,0),1)
```

```python
import keras
from keras.models import Sequential
from keras.layers import Dense, Input
from sklearn.model_selection import KFold
import time
import numpy as np

#Error cuadrático medio (para problemas de regresión)
def ECM(Y_real,Y_est):
    N = np.size(Y_real)
    ecm = sum((Y_est - Y_real)**2)/N
    return ecm

def crear_modelo_reg(input_dim, neuronas):
    model = Sequential()
    model.add(Input(shape=(input_dim,)))
    model.add(Dense(neuronas, activation='relu'))  # Capa oculta
    model.add(Dense(1, activation='linear'))       # Capa de salida: función identidad
    model.compile(loss='mean_absolute_error', optimizer='adam', metrics=['mean_squared_error'])
    return model

def evaluar_regresion(X, Y, epocas, neuronas, verbose=0):
    #Para calcular el costo computacional
    tiempo_i = time.time()

    #Implementamos la metodología de validación cross validation con 10 folds
    Errores = np.ones(10)
    j = 0
    kf = KFold(n_splits=10)

    # Guardamos las predicciones de cada fold para poder graficarlas después
    y_real_total = np.zeros_like(Y)
    y_pred_total = np.zeros_like(Y)

    for train_index, test_index in kf.split(X):
        X_train, X_test = X[train_index], X[test_index]
        y_train, y_test = Y[train_index], Y[test_index]

        # Modelo nuevo en cada fold (pesos reiniciados), compilado con el optimizador Adam
        model = crear_modelo_reg(X.shape[1], neuronas)

        # Entrenamiento del modelo
        model.fit(X_train, y_train, epochs=epocas, batch_size=100, validation_split=0.1, verbose=verbose)

        # Validación del modelo
        y_pred = model.predict(X_test, verbose=0)

        Errores[j] = ECM(y_test, y_pred)[0]
        y_real_total[test_index] = y_test
        y_pred_total[test_index] = y_pred
        j+=1

    t = time.time()-tiempo_i
    print(f"Épocas={epocas}, neuronas en capa oculta={neuronas}")
    print("Error de validación: " + str(np.mean(Errores)) + " +/- " + str(np.std(Errores)))
    print("Tiempo total de ejecución: " + str(t) + " segundos.\n")
    return np.mean(Errores), np.std(Errores), t, y_real_total, y_pred_total

# Ejecución individual (cambie los parámetros según la fila de la tabla)
_ = evaluar_regresion(X, Y, epocas=100, neuronas=10)
```

3.1 Defina en sus palabras que es una época en el modelo de redes neuronales artificiales.

R/: Una época es una pasada completa de todos los datos de entrenamiento por la red. En una época cada muestra se presenta una vez a la red, se calcula el error y se actualizan los pesos (en este caso, por lotes de 100 muestras con el optimizador Adam). Entrenar durante varias épocas significa repetir ese proceso muchas veces para que la red vaya reduciendo el error poco a poco.

3.2 Por qué se usa la función de activación lineal en lugar de reLu, tanh o sigmoid en la capa de salida para este problema?

R/: Porque es un problema de regresión: la salida es un valor continuo (la humedad absoluta multiplicada por 10) que puede tomar cualquier valor real, sin un rango fijo. La función lineal (identidad) deja pasar el valor calculado por la neurona sin modificarlo, por lo que la red puede predecir cualquier número. Las otras funciones limitan el rango de la salida: ReLU elimina los valores negativos, la sigmoide solo produce valores entre 0 y 1 y tanh entre -1 y 1, por lo que no podrían representar valores como 8 o 15. Esas funciones sí son útiles en las capas ocultas, donde se necesita introducir no linealidad, pero la activación de la capa de salida debe ser compatible con el rango de la variable que se quiere predecir.

3.4 Para esta base de datos con casi 10 mil registros, que puede decir del costo computacional del modelo con 100 épocas y 10 neuronas en la capa oculta

R/: El costo computacional es considerable para una red tan pequeña: 305.8 s (aprox. 5 minutos) frente a 53.4 s con 10 épocas. Con 10 veces más épocas el tiempo aumenta cerca de 5.7 veces (no exactamente 10) porque parte del tiempo de cada fold es fijo. El costo es alto porque en cada época la red recorre todo el conjunto de entrenamiento (cerca de 9 mil registros en lotes de 100) y, además, la validación cruzada de 10 folds repite el entrenamiento 10 veces, así que en total se ejecutan 1000 épocas.

A cambio, el error baja de forma importante: el ECM pasa de 246.11 a 68.69 (una reducción de cerca del 72 %). Pero el costo se paga con una ganancia decreciente. Con 10 épocas y 20 neuronas se obtuvo un ECM de 115.49 en el mismo tiempo que con 10 neuronas (53.5 s), es decir, aumentar las neuronas casi no cuesta tiempo en una red pequeña. Por eso sería más eficiente probar más neuronas con un número moderado de épocas, o usar *early stopping* para detener el entrenamiento cuando el error deje de mejorar. En redes más grandes o con más datos, este costo crecería mucho más.
